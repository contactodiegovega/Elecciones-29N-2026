```python
import pandas as pd
from pathlib import Path
PROC = Path('../data/processed')
turnout = pd.read_parquet(PROC / 'election_turnout.parquet')
results = pd.read_parquet(PROC / 'election_results.parquet')
elections = pd.read_parquet(PROC / 'elections.parquet')
print(elections[['election_date' if 'election_date' in elections else elections.columns[0]]].to_string())
print('turnout:', turnout.shape, '| results:', results.shape)
```

```python
nat = turnout.groupby('election_date', as_index=False)[['electores','votantes']].sum()
nat['participacion'] = nat['votantes'] / nat['electores'] * 100
ax = nat.plot(x='election_date', y='participacion', marker='o', legend=False)
ax.set_ylabel('% participacion nacional'); ax.set_title('Congreso 1977-2023 (datos: Interior)')
```
